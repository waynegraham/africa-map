```python
from pathlib import Path
import json
import pandas as pd
from heritage_flood import national_context

ROOT = Path.cwd()
if ROOT.name == 'notebooks':
    ROOT = ROOT.parent
assert (ROOT / 'config/national_context.json').exists()
ACQUIRE_CONTEXT = False
ACQUIRE_SCENARIOS = False
ACQUIRE_ACCESS = False
REBUILD = False
if ACQUIRE_CONTEXT:
    national_context.acquire(ROOT)
if ACQUIRE_SCENARIOS:
    national_context.acquire_scenarios(ROOT)
if ACQUIRE_ACCESS:
    national_context.acquire_access(ROOT)
if REBUILD:
    national_context.build(ROOT)
```

```python
out = ROOT / 'data/processed/national_context'
context = json.loads((out / 'context.json').read_text())
pd.Series(context['coverage'], name='queried source records')
```
